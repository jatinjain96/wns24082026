##Creating dataframe from different sources

```python
dfr=spark.read
df = dfr.csv(path='/Volumes/wns24082026/quickstart_schema/sandbox/datasets/sample_dataset/users.csv',
header = True,
inferSchema = True)

display(df)
```

Create dataframe from JSON

```python
df = dfr.parquet('/Volumes/wns24082026/quickstart_schema/sandbox/datasets/sample_dataset/users.parquet')
display(df)
```

####Attaching custom schema

```python
from pyspark.sql.types import *

USER_SCHEMA = StructType(
    [
        StructField("id",IntegerType()),
        StructField("age",IntegerType()),
        StructField("gender",StringType()),
        StructField("designation",StringType()),
        StructField("salary",IntegerType()),
    ]
)

df = spark.read.csv(path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/sample_dataset/users.csv",
header = True,
schema = USER_SCHEMA)

display(df)
```

####Handling Bad Records
1) Permissive (Default)
2) Dropmalformed
3) Failfast

```python
df = spark.read.option('columnNameOfCorruptRecord','bad_record').json(path='/Volumes/wns24082026/quickstart_schema/sandbox/datasets/access_logs.json',
mode='FAILFAST')
display(df)
```

####Dataframe Writer API

1) Error if exists
2) Overwrite
3) Append
4) Ignore

```python
df = spark.read.csv(path='/Volumes/wns24082026/quickstart_schema/sandbox/datasets/sample_dataset/users.csv',
header = True,
inferSchema = True)

df.write.save(
    format='PARQUET',
    path='/Volumes/wns24082026/quickstart_schema/sandbox/output/output_parquet',
    mode='append'
)
```
