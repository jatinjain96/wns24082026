```python
dbutils.widgets.dropdown(
     name="exp_level",
    defaultValue="Mid Level",
    choices=["Invalid Experience","Mid Level","Junior","Senior"],
    label="Experience Level",
)
 
```

```python
from pyspark.sql.functions import when, col
df = spark.read.csv(path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/employee.csv",
                    header=True,
                    inferSchema=True,
                    sep="|",
                    quote="'")
 
result_df = df.withColumn(
    "category",
    when(col("exp") >= 10, "Senior")
    .when(col("exp") >= 5, "Mid Level")
    .when(col("exp") > 0, "Junior")
    .otherwise("Invalid Experience"),
)

exp_level = dbutils.widgets.get(name="exp_level")
result_df.groupBy("category").count().sort("count", ascending=False).filter(col("category")==exp_level).display()
 
```
