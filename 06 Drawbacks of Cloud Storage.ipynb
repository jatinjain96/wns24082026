###Prerequisties

####Load users_001.csv into Database

```python
df = spark.read.csv('/Volumes/wns24082026/quickstart_schema/sandbox/datasets/user_dataset/users_001.csv',
                    header=True,
                    inferSchema=True)
```

```python
df.write.save(
    format="PARQUET",
    path="/Volumes/wns24082026/quickstart_schema/sandbox/output_parquet",
)
```

```python
df_parquet = spark.read.parquet("/Volumes/wns24082026/quickstart_schema/sandbox/output_parquet")
df_parquet.display()
```

```python
df_parquet.createOrReplaceTempView("users_vw")
```

###Disadvantage - Updates not allowed

```python
%sql

UPDATE users_vw
SET country = 'Bharat'
WHERE country = 'India'
```

###Overwrite is allowed, No rollback - version history

```python
from pyspark.sql.functions import *

df.filter(col("country") == "India").write.save(
    format="PARQUET",
    path="/Volumes/wns24082026/quickstart_schema/sandbox/output_parquet",
    mode="OVERWRITE",
)
```
