###Derive Columns

```python
from pyspark.sql.functions import *

df.withColumn("category", lit("not defined")).show()
```

```python
result_df = df.withColumn("category", 
              when(col('exp') >= 10, "Senior")
              .when(col('exp') >= 5, "Mid level")
              .when(col('exp') > 0, "Junior")
              .otherwise("Invalid Experience"))

result_df.select("exp","category").show()
```

```python
result_df.groupBy('category').count().sort('count').display()
```

```python
result_df.groupBy("category").agg(min("exp"),
                                  max("exp"),
                                  avg("exp")).show()
```
